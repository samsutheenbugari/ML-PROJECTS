```python
lpp['Sub_category']=lpp['Sub_category'].str.strip()    # remove spaces
```

```python
new_row = [{'Sub_category': 'UNK', 'Hp': 0.065, 'Dell': 0.035},{'Sub_category': 'SJ', 'Hp': 0.065, 'Dell': 0.035}]
lpp=pd.concat([lpp, pd.DataFrame(new_row)], ignore_index=True)  #adding new row 
```

```python
lpp['Sub_category'] = lpp['Sub_category'].str.replace('Toner', 'TON')
lpp['Sub_category'] = lpp['Sub_category'].str.replace('Laserjet', 'LJ')
lpp['Sub_category'] = lpp['Sub_category'].str.replace('Inkjet', 'IJ')
lpp['Sub_category'] = lpp['Sub_category'].str.replace('Ink', 'INK')
#replace string values it should perform one by one
```

```python
lpp
```

```python
keys=lpp['Sub_category'].tolist()
values1=lpp['Hp'].tolist()
values2=lpp['Dell'].tolist()	
#list conversion
```

```python
#converts list to dictionary using zip function
replacements1=dict(zip(keys, values1))
replacements2=dict(zip(keys, values2))
```

```python
replacements1
```

```python
replacements =replacements1                           #filling the llp detection values to the dataframe

for item, value in replacements.items():
    sdf1.loc[
        (sdf1['Item'] == item) & 
        (sdf1['Brand'].str.contains('HP', na=False)), 
        'lp'
    ] = sdf1.loc[
        (sdf1['Item'] == item) & 
        (sdf1['Brand'].str.contains('HP', na=False)), 
        'lp'
    ].replace(0, value)


```

```python
replacements = replacements2
for item, value in replacements.items():
    sdf1.loc[
        (sdf1['Item'] == item) & 
        (sdf1['Brand'].str.contains('DELL', na=False)), 
        'lp'
    ] = sdf1.loc[
        (sdf1['Item'] == item) & 
        (sdf1['Brand'].str.contains('DELL', na=False)), 
        'lp'
    ].replace(0, value)
```
