```python

url=laptop1 #linkurl

headers={"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/128.0.0.0 Safari/537.36 Edg/128.0.0.0"}
#header used to save some website blocks

page=requests.get(url,headers=headers) #page request to link url
```

```python
soup=BeautifulSoup(page.content,"html.parser")   #get content
soup=BeautifulSoup(soup.prettify(),"html.parser") #shape the page html content
```

#Geting info
soup.find().get_text()

```python
#get title using find function and convert into text

title=soup.find(id='productTitle').get_text()
title=title.strip()
title
```

```python
#getting seller name with id tag

seller_name=soup.find(id="sellerProfileTriggerId").get_text()
seller_name=seller_name.strip()
seller_name
```

```python
#get price

price=soup.find(id="corePriceDisplay_desktop_feature_div").get_text()
price=price.strip()[1:7]
price = ''.join(char for char in price if char.isdigit())
price=int(price)
price 
```
